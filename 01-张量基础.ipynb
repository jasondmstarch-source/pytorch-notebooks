# 01 · 张量基础

PyTorch 里所有数据都是 **张量（Tensor）**。你可以先把它理解成：

> 一个「能做自动求导、能上 GPU 的 NumPy 数组」。

光电、图像、光谱数据，最后都会变成张量喂进模型。所以这一节先把最基础的操作摸熟。

```python
import torch

print('PyTorch 版本:', torch.__version__)
print('可用线程数  :', torch.get_num_threads())
```

## 1. 创建张量

有几种常用造法，先都试一遍。

```python
# 从 Python 列表创建
a = torch.tensor([1, 2, 3])
print('一维张量:', a)

# 二维（就是矩阵）
b = torch.tensor([[1, 2],
                  [3, 4]])
print('二维张量:')
print(b)
```

```python
# 特殊矩阵
print('全 0:')
print(torch.zeros(2, 3))

print('全 1:')
print(torch.ones(2, 3))

print('单位矩阵:')
print(torch.eye(3))

# 随机数 —— 深度学习里初始化权重就靠它
print('均匀分布 rand(0~1):')
print(torch.rand(2, 3))

print('标准正态分布 randn:')
print(torch.randn(2, 3))

# 等差数列，很像 Python 的 range
print('arange(0, 10, 2):', torch.arange(0, 10, 2))
```

## 2. 张量的「形状」—— 最重要的概念

深度学习里 90% 的报错都是**形状对不上**。所以每次写代码，心里要清楚每个张量是什么形状。

```python
x = torch.randn(2, 3, 4)

print('张量 x:')
print(x)
print()
print('形状 shape :', x.shape)    # torch.Size([2, 3, 4])
print('维度 ndim  :', x.ndim)     # 3 维
print('元素个数   :', x.numel())  # 2*3*4 = 24
print('数据类型   :', x.dtype)
```

```python
# reshape：改变形状，但元素总数不变
flat = torch.arange(12)
print('原始:', flat)

print('\nreshape(3, 4):')
print(flat.reshape(3, 4))

# -1 表示「这一维你自己算」
print('\nreshape(3, -1) 效果一样:')
print(flat.reshape(3, -1))

# 图像数据常用：把 (H, W) 拉平成一根向量
print('\nreshape(-1) 拉平:', flat.reshape(3, 4).reshape(-1))
```

## 3. 运算

⚠️ **新手最容易搞混的一点**：
- `*` 是**逐元素相乘**（对应位置各乘各的）
- `@` 才是**矩阵乘法**

```python
x = torch.tensor([[1., 2.],
                  [3., 4.]])
y = torch.tensor([[10., 20.],
                  [30., 40.]])

print('x:')
print(x)
print('\ny:')
print(y)

print('\n加法 x + y:')
print(x + y)

print('\n逐元素乘 x * y（1*10, 2*20, 3*30, 4*40）:')
print(x * y)

print('\n矩阵乘 x @ y（行乘列再求和）:')
print(x @ y)
```

```python
# 广播（broadcasting）：形状不一样也能运算，PyTorch 自动扩展
x = torch.tensor([[1., 2.],
                  [3., 4.]])

print('x + 10（标量自动广播到每个元素）:')
print(x + 10)

print('\nx * 列向量 [[1],[100]]:')
print(x * torch.tensor([[1.], [100.]]))
```

## 4. 索引与切片

和 NumPy 几乎一模一样。

```python
x = torch.arange(12).reshape(3, 4)
print('x:')
print(x)

print('\n第 0 行 x[0]      :', x[0])
print('第 1 列 x[:, 1]   :', x[:, 1])
print('\n子块 x[0:2, 1:3]:')
print(x[0:2, 1:3])

# 按条件筛选，做数据清洗时很实用
print('\nx > 5 的元素:', x[x > 5])
```

## 5. 和 NumPy 互通

实际项目里经常用 NumPy 读数据、用 PyTorch 算模型，两者可以自由转换。

```python
import numpy as np

n = np.array([1., 2., 3.])
t = torch.from_numpy(n)

print('numpy -> torch:', t)
print('torch -> numpy:', t.numpy())

# ⚠️ 注意：默认情况下两者「共享内存」
t[0] = 99
print('\n改了 torch 的 t[0] 之后，numpy 也跟着变了:', n)
```

```python
# 想要互不影响，就 clone 一份
n = np.array([1., 2., 3.])
t = torch.from_numpy(n).clone()

t[0] = -1
print('clone 之后改 torch，numpy 不受影响:', n)
```
